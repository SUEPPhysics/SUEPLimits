# Significances

Plot a scan of significances.

```python
import os
import glob
import uproot
import matplotlib.pyplot as plt
import plot_utils
from prettytable import PrettyTable
```

```python
motherDir = "../cards/WH_12_10_gamma_v2/"
subDirs = glob.glob(motherDir + "cards-*")
```

```python
significances = []
for subDir in subDirs:
    _f = uproot.open(subDir + "/higgsCombineTest.Significance.mH120.root")
    _significance = _f['limit']['limit'].array()
    _sample_params = list(plot_utils.get_params_from_sample_name_wh(subDir))
    _sample_params[-1] = plot_utils.mA[_sample_params[-1]]
    significances.append((_sample_params, _significance[0]))
```

```python
table = PrettyTable()

param_names = ['\mS [\GeV]', '\mPhi [\GeV]', '\TD [\GeV]', "\mAprime [\GeV]", 'Significance']
table.field_names = param_names

for params, significance in significances:
    row = list(params) + [round(significance, 3)]
    table.add_row(row)

print(table)
```
